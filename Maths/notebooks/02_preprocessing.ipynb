```python
import pandas as pd
import matplotlib.pyplot as plt

# Load raw and processed datasets
raw = pd.read_csv(
    "../data/synthetic/synthetic_fmri.csv"
)

processed = pd.read_csv(
    "../data/processed/synthetic_fmri_preprocessed.csv"
)

roi_cols = [
    col for col in processed.columns
    if col.startswith("ROI_")
]

# Check shape
print("Raw shape:", raw.shape)
print("Processed shape:", processed.shape)
```

```python
# Check normalized means
print("\nROI Means:")
print(processed[roi_cols].mean().round(4))
```

```python
# Check normalized standard deviations
print("\nROI Standard Deviations:")
print(processed[roi_cols].std(ddof=0).round(4))
```

```python
plt.figure(figsize=(14, 7))

plt.subplot(2, 1, 1)

for col in roi_cols[:4]:
    plt.plot(raw["time_seconds"], raw[col], label=col)

plt.title("Raw fMRI Signals")
plt.ylabel("BOLD Signal")
plt.legend()
plt.grid()

plt.subplot(2, 1, 2)

for col in roi_cols[:4]:
    plt.plot(
        processed["time_seconds"],
        processed[col],
        label=col
    )

plt.title("Preprocessed fMRI Signals")
plt.xlabel("Time (seconds)")
plt.ylabel("Normalized Signal")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```
